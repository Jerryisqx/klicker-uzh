```python
import pandas as pd

file_path = "/content/MP_Export_07-03_ElementGenerated(in).xlsx"
df = pd.read_excel(file_path)

df.head()
```

```python
score_columns = ["ambiguity", "difficultyRate", "factualCorrectness", "relevance", "typeRate"]
descriptive_stats = df[score_columns].describe()
import pandas as pd

print(descriptive_stats)
# import ace_tools as tools
# tools.display_dataframe_to_user(name="Descriptive Statistics", dataframe=descriptive_stats)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Computed correlation matrix
correlation_matrix = df[score_columns].corr()

# Mapping heat map
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Evaluation Metrics")
plt.show()
```

```python
# Plot the distribution histogram of each score indicator
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for i, col in enumerate(score_columns):
    sns.histplot(df[col], bins=10, kde=True, ax=axes[i])
    axes[i].set_title(f"Distribution of {col}")
    axes[i].set_xlabel(col)
    axes[i].set_ylabel("Count")

plt.tight_layout()
plt.show()

```
